# `geo4bolivia.load`

The example of [`load`](https://quarcs-lab.github.io/geo4bolivia/reference/load.html) in the [geo4bolivia function reference](https://quarcs-lab.github.io/geo4bolivia/reference/). Run the cell; the reference page documents every argument.

```python
!pip install -q geo4bolivia
import geo4bolivia as geo
import plotly.express as px

df = geo.load("sdg05", variables=["index_sdg5"], years=[2012, 2024],
              groups=["department"])
px.histogram(df, x="index_sdg5", color="year", barmode="overlay",
             labels=geo.labels(df), title="Gender equality index")
```
