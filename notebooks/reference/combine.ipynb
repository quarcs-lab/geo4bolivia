# `geo4bolivia.combine`

The example of [`combine`](https://quarcs-lab.github.io/geo4bolivia/reference/combine.html) in the [geo4bolivia function reference](https://quarcs-lab.github.io/geo4bolivia/reference/). Run the cell; the reference page documents every argument.

```python
!pip install -q geo4bolivia
import geo4bolivia as geo
import plotly.express as px

dep = geo.combine(["sdg03", "sdg04"], scale="department", years=2024)
dep = geo.add_names(dep)
px.scatter(dep, x="index_sdg3", y="index_sdg4", text="name",
           labels=geo.labels(dep), title="Health and education, ~2024")
```
