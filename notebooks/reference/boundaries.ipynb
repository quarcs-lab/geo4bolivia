# `geo4bolivia.boundaries`

The example of [`boundaries`](https://quarcs-lab.github.io/geo4bolivia/reference/boundaries.html) in the [geo4bolivia function reference](https://quarcs-lab.github.io/geo4bolivia/reference/). Run the cell; the reference page documents every argument.

```python
!pip install -q geo4bolivia
import geo4bolivia as geo
import plotly.express as px

dep = geo.boundaries("department")
px.choropleth_map(dep, geojson=dep.geometry, locations=dep.index,
                  color="name", hover_name="name",
                  center={"lat": -16.3, "lon": -64}, zoom=4,
                  title="Bolivia's nine departments")
```
