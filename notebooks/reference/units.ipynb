# `geo4bolivia.units`

The example of [`units`](https://quarcs-lab.github.io/geo4bolivia/reference/units.html) in the [geo4bolivia function reference](https://quarcs-lab.github.io/geo4bolivia/reference/). Run the cell; the reference page documents every argument.

```python
!pip install -q geo4bolivia
import geo4bolivia as geo
import plotly.express as px

m = geo.units("municipality")              # names, parents, INE codes, areas
display(m[["geo_id", "name", "ine_code", "province_id", "department_id"]].head())
px.histogram(m, x="area_km2", log_y=True, title="Municipal areas (km²)")
```
