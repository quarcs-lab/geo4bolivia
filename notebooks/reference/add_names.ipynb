# `geo4bolivia.add_names`

The example of [`add_names`](https://quarcs-lab.github.io/geo4bolivia/reference/add_names.html) in the [geo4bolivia function reference](https://quarcs-lab.github.io/geo4bolivia/reference/). Run the cell; the reference page documents every argument.

```python
!pip install -q geo4bolivia
import geo4bolivia as geo
import plotly.express as px

df = geo.load("sdg08", variables="index_sdg8", years=2024)
df = geo.add_names(df, groups=["department", "province"])
px.box(df, x="department", y="index_sdg8", hover_name="name",
       labels=geo.labels(df), title="Decent work index, ~2024")
```
