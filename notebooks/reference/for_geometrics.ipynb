# `geo4bolivia.for_geometrics`

The example of [`for_geometrics`](https://quarcs-lab.github.io/geo4bolivia/reference/for_geometrics.html) in the [geo4bolivia function reference](https://quarcs-lab.github.io/geo4bolivia/reference/). Run the cell; the reference page documents every argument.

```python
!pip install -q geo4bolivia geometrics
import geo4bolivia as geo
import geometrics as gm

gdf, df, df_dict = geo.for_geometrics("sdg", variables="index_sdg")
df = gm.set_labels(df, df_dict, set_panel=True)
w = gm.make_weights(gdf, method="queen")
gm.explore_moran_plot(df, "index_sdg", gdf=gdf, w=w, period=2024).fig
```
